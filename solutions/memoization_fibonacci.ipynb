```python
import time
import matplotlib.pyplot as plt

# Part 1: recursive Fibonacci, no memoization
def fibonacci1(n):
    if n == 1 or n == 2:
        return 1
    else:
        return fibonacci1(n-1) + fibonacci1(n-2)

times = []
for n in range(1, 26):
    start_time = time.time()
    fibonacci1(n)
    times.append(time.time() - start_time)
    
fig, ax1 = plt.subplots()
ax1.plot(range(1, 26), times, '-o', color='tab:red', label='Linear scale')
ax1.set_xlabel('n')
ax1.set_ylabel('Time (seconds)', color='tab:red')
ax1.tick_params(axis='y', labelcolor='tab:red')

ax2 = ax1.twinx()
ax2.semilogy(range(1, 26), times, '-s', color='tab:blue', label='Log scale')
ax2.set_ylabel('Time (log scale)', color='tab:blue')
ax2.tick_params(axis='y', labelcolor='tab:blue')

fig.tight_layout()
plt.show()
```
